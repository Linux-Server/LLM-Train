```python
from langchain.agents import create_agent
from dotenv import load_dotenv

load_dotenv()

agent = create_agent(
    model="gpt-5-nano"
)
```

```python
agent.invoke({"messages" :"heloo"})
```

```python
### Create a sub agent
from langchain.tools import tool

sub_agent = create_agent(
    model="gpt-5-nano"
)

@tool(name_or_callable="research", description="Tell a joke on the topic")
def my_worker(query:str):
    result = sub_agent.invoke({"messages": query})
    return result["messages"][-1].content
```

```python
main_agent = create_agent(
    model="gpt-5-nano",
    tools=[my_worker]
)
```

```python
main_agent.invoke({"messages": "Artificial Intelligence"})
```
