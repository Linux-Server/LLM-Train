```python
from deepeval.synthesizer import Synthesizer
from deepeval.synthesizer.config import StylingConfig

styling_config = StylingConfig(
    input_format="""
Everyday questions asked by normal people about life, work, learning,
technology, emotions, productivity, or confusion.

The user may sound stressed, curious, unsure, or excited.
Questions must feel natural and human.
""",

    expected_output_format="""
A warm, gentle, and polite assistant response.

Response rules:
- Kind and respectful tone
- Acknowledge the user's feeling or intention
- Clear and simple explanation
- Encourage and reassure the user
- Avoid robotic or overly formal language
- No lecturing or judging
- Short paragraphs
- Optional helpful suggestion at the end
""",

    task="""
Provide supportive and helpful responses like a thoughtful human assistant.
""",

    scenario="""
A friendly personal AI assistant helping people in daily life.
The assistant behaves like a patient teacher + supportive friend.
"""
)

synthesizer = Synthesizer(styling_config=styling_config)
```

```python
from deepeval.synthesizer import Synthesizer

goldens = synthesizer.generate_goldens_from_scratch(num_goldens=5)
print(goldens)
```

```python
goldens = synthesizer.generate_goldens_from_goldens(
  goldens=goldens,
  max_goldens_per_golden=2,
  include_expected_output=True,
)
```
