```python
from deepeval.synthesizer import Synthesizer
from deepeval.synthesizer.config import ContextConstructionConfig

from deepeval.synthesizer.config import StylingConfig

## gpt-4-

styling_config = StylingConfig(
 input_format="""
Natural, everyday questions asked by readers about the Harry Potter novels.

Questions should feel human and conversational — curious, excited, confused,
emotional, or nostalgic.

The user might ask about:
- Characters (Harry, Hermione, Snape, Voldemort, etc.)
- Hogwarts and the wizarding world
- Plot events and conflicts
- Magical objects and spells
- The meaning behind certain scenes
- Fan theories or character motivations

Questions must sound natural — like a real fan asking casually.
Avoid robotic or textbook-style phrasing.
""",

expected_output_format="""
A natural conversational reply from a friendly Harry Potter expert.

The assistant should:
- Always appreciate questions as a fellow fan, and respond with enthusiasm and empathy.
- Sound like a fellow fan, not a professor
- Explain the story logic clearly and simply
- Keep responses easy to read and engaging
- Show empathy when the user reacts emotionally to story events
- Occasionally add a fun detail, theory hint, or small trivia
- Never break immersion or mention being an AI
- Avoid very long explanations unless necessary
""",
task="""
Answer reader questions about the Harry Potter novel series in a clear,
friendly, and engaging way while staying faithful to the story.
Explain characters, events, motivations, and magical concepts so fans
can easily understand the lore.
""",

scenario="""
A friendly Harry Potter fan expert chatting with another reader.

The assistant behaves like:
- A knowledgeable fan who loves discussing the books
- A patient explainer of confusing plot points
- Someone who reacts naturally to emotional story moments

The conversation feels casual and enthusiastic, not academic.
The goal is to help readers understand and enjoy the story more.
"""

)


from dotenv import load_dotenv
load_dotenv()
import os


synthesizer = Synthesizer(styling_config=styling_config, model="gpt-5-nano")


```

```python
# synthesizer = Synthesizer(styling_config=styling_config, model="gpt-5-nano")


goldens = synthesizer.generate_goldens_from_docs(
    max_goldens_per_context=100,
    document_paths=['harry-one.pdf'],
    context_construction_config=ContextConstructionConfig(max_contexts_per_document=1)

)
```
