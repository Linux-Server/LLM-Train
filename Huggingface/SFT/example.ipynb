```python
model_name = "Qwen/Qwen3-0.6B-Base"
from datasets import load_dataset

dataset = load_dataset("trl-lib/Capybara",)
```

```python
from trl import SFTTrainer, SFTConfig

trainer = SFTTrainer(
    model=model_name,
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    args = SFTConfig(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        learning_rate=2e-4,
        output_dir="./Qwen-0.6B-Capybara",
        overwrite_output_dir=True,
        # num_train_epochs=1,
        max_steps = 120,
        save_total_limit=2,
        lr_scheduler_type="linear",
        logging_steps=20,
        report_to="none",
        chat_template_path= "Qwen/Qwen3-0.6B",
        eval_strategy="steps",
        eval_steps=20
    )
)
```

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
n_model = AutoModelForCausalLM.from_pretrained("./Qwen-0.6B-Capybara/checkpoint-120", device_map="auto")
n_tokenizer = AutoTokenizer.from_pretrained("./Qwen-0.6B-Capybara/checkpoint-120", fix_mistral_regex=True)
```

```python
prompt = [
    {"role": "user", "content": "Who are you?"}]

model_inps = n_tokenizer.apply_chat_template(prompt, tokenize=False, add_generation_prompt=True, enable_thinking=True)
model_inps

model_inps = n_tokenizer(model_inps, return_tensors="pt").to("cuda")
model_inps

out = n_model.generate(**model_inps, max_new_tokens=100)

n_tokenizer.batch_decode(out)
```

```python
model_name = "Qwen/Qwen3-0.6B-Base"
from datasets import load_dataset

dataset = load_dataset("trl-lib/Capybara",)

from trl import SFTTrainer, SFTConfig

trainer = SFTTrainer(
    model=model_name,
    train_dataset=dataset["train"],
    eval_dataset=dataset["test"],
    args = SFTConfig(
        per_device_train_batch_size=2,
        gradient_accumulation_steps=4,
        learning_rate=2e-4,
        output_dir="./Qwen-0.6B-Capybara",
        overwrite_output_dir=True,
        # num_train_epochs=1,
        max_steps = 120,
        save_total_limit=2,
        lr_scheduler_type="linear",
        logging_steps=20,
        report_to="none",
        chat_template_path= "Qwen/Qwen3-0.6B",
        eval_strategy="steps",
        eval_steps=20
    )
)

stats = trainer.train()
```
